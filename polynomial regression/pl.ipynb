```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X=6*np.random.rand(100,1)-3
y=0.5*X**2+1.5*X+2+np.random.randn(100,1)
plt.scatter(X,y,color='g')
plt.xlabel('X dataset')
plt.ylabel('y dataset')
plt.title('Polynomial Regression Data')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,model.predict(X_test))
print(score)
```

```python
plt.plot(X_train,model.predict(X_train),color='r')
plt.scatter(X_train,y_train,color='g')
plt.xlabel('X dataset')
plt.ylabel('y dataset')
plt.title('Polynomial Regression Data')
plt.show()
```

```python
#polynomial reg
#Lets apply polynomial transformation to our data
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2,include_bias=True)
X_train_poly=poly.fit_transform(X_train)
X_test_poly=poly.transform(X_test)

```

```python
X_train_poly
```

```python
model_2=LinearRegression()
model_2.fit(X_train_poly,y_train)
y_pred=model_2.predict(X_test_poly)
score=r2_score(y_test,y_pred)
print(score)
```

```python
print(model_2.coef_)
```

```python
model_2.intercept_
```

```python
plt.scatter(X_train,model_2.predict(X_train_poly),color='r')
plt.scatter(X_train,y_train,color='g')
plt.xlabel('X dataset')
plt.ylabel('y dataset')
plt.title('Polynomial Regression Data')
plt.show()
```

```python
#Prediction of new data
X_new=np.linspace(-3,3,200).reshape(200,1)
X_new_poly=poly.transform(X_new)
y_new=model_2.predict(X_new_poly)
plt.plot(X_new,y_new,color='r',linewidth=2,label=' New Predictions')
plt.scatter(X_train,y_train,color='b',label='Training points')
plt.scatter(X_test,y_test,color='g',label='Testing points')
plt.xlabel('X dataset')
plt.ylabel('y dataset')
plt.legend()
plt.show()

```
