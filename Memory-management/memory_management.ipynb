Memory management best practices
1.Use of local variables
2.Explicitly delete objects
3.Avoid cicular references
4.Profile Mmeory usage


```python
import gc 
class myObj:
    def __init__(self,name):
        self.name=name
        print(f"Object {self.name} created")
    def __del__(self):
        print(f"Object {self.name} deleted")

#Creating circular reference
obj1=myObj('Obj1')
obj2=myObj('Obj2')

obj1.ref=obj2
obj2.ref=obj1

del obj1
del obj2

#Manual trigger gc
gc.collect()
```

```python
#Genertor for Memory efficiency
#Generators allow you to produce items one at a time using memory efficiently by only keeping on item in the memory at a time

def generate_numbeers(n):
    for i in range(n):
        yield i

#ising generator
for num in generate_numbeers(100000):
    print(num)
    if num>10:
        break

```

```python
#Profiling memory usage using tracemalloc
import tracemalloc
def create_list():
    return [i for i in range(10000)]
def main():
    tracemalloc.start()
    create_list()

    snapshots=tracemalloc.take_snapshot()
    top_stats=snapshots.statistics('lineno')

    print("[Top 10]")
    for stat in top_stats[:10]:
        print(stat)
        
```

```python
main()
```
