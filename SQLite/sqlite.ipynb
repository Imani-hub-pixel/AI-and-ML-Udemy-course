```python
import sqlite3
```

```python
#Connect to a sqlite database
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
#create table
cursor.execute('''
create table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department Text
    )
''')

#commit the changes

connection.commit()
```

```python
cursor.execute('''
select*from employees
               ''')
```

```python
#Insert the data in the sqlite table
cursor.execute('''
INSERT INTO employees(name,age,department)
               values('faith',20,'Software developer')
''')

cursor.execute('''
INSERT INTO employees(name,age,department)
    VALUES('Bob',22,' Engineering')
''')

cursor.execute('''
INSERT INTO employees(name,age,department)
               values('Mercy',25,' Medicine')
''')
connection.commit()
```

```python
#Query the data
cursor.execute('Select*from employees')
rows=cursor.fetchall()
for row in rows:
    print(row)
```

```python
#Update the data in the table
cursor.execute('''

UPDATE employees
    set age=34
    where name='faith'
                   
''')
connection.commit()
```

```python
cursor.execute('select*from employees')
rows=cursor.fetchall()
for row in rows:
    print(row)
```

```python
#Delete the data from the table
cursor.execute('''
DELETE from employees
               where name='Bob'
''')
```

```python
cursor.execute('select*from employees')
rows=cursor.fetchall()
for row in rows:
    print(row)
```

```python
#Working wit sales data
connection=sqlite3.connect('Coffee_sales.db')
cursor=connection.cursor()

#create a table
cursor.execute('''
CREATE TABLE IF NOT EXISTS  sales(
               id INTERGER PRIMARY KEY,
               date TEXT NOT NULL,
               product TEXT NOT NULL,
               sales INTERGER,
               region TEXT
               )
               
''')



```

```python
#Insert data
coffee_sale=[
    ('2023-01-01','product1',100,'North'),
    ('2023-01-02','product2',200,'South'),
    ('2023-01-03','product1',150,'East'),
    ('2023-01-04','product3',250,'West'),
    ('2023-01-05','product2',300,'North')
    
]

connection.commit()

cursor.executemany('''
Insert into sales (date,product,sales,region)
                   values(?,?,?,?)

''',coffee_sale)

connection.commit()
```

```python
cursor.execute('select*from sales')
rows=cursor.fetchall()
for row in rows:
    print(row)
```
