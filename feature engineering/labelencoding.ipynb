```python
lbl_encoder.fit_transform(df['color'])
```

```python
#Ordinal encoding is not a good choice for nominal categorical variables. 
# It can lead to the model interpreting the encoded values as having an inherent order or hierarchy, which may not be the case. 
# This can result in inaccurate predictions and reduced model performance.
#  For nominal categorical variables, one-hot encoding is often a better choice as 
# it creates binary columns for each category, 
# avoiding any unintended ordinal relationships.
```

```python
#Ordinal encoding-assigning ranks to the categories i.e  degree:masters,bachelors,phd,highschool etc
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    'degree':['masters','bachelors','phd','highschool','masters','bachelors','phd']
})
```

```python
#Assign ranks to the categories in the 'degree' column
ordinal_encoder = OrdinalEncoder(categories=[['highschool', 'bachelors', 'masters', 'phd']])
df['degree_encoded'] = ordinal_encoder.fit_transform(df[['degree']])
df['degree_encoded']
```

```python
#Target guided ordinal encoding-assigning ranks to the categories based on the target variable.
```

```python
city_df=pd.DataFrame({
    "city":['cityA','cityB','cityC','cityA','cityB','cityC'],
    "price":[100,200,300,150,250,350]
})
```

```python
mean_price=city_df.groupby('city')['price'].mean().to_dict()
```

```python
df['city_encoded']=city_df['city'].map(mean_price)
df['city_encoded']
```

```python
import seaborn as sns
tips_df1=sns.load_dataset('tips').head()
```

```python
encode_tips_df1=pd.DataFrame({
    "total_bill":tips_df1['total_bill'],
    "day":tips_df1['day']
})
```

```python
total_bill_mean=encode_tips_df1.groupby('day')['total_bill'].mean().to_dict()
```

```python
tips_df1['day_encoded'] = tips_df1['day'].map(total_bill_mean)
tips_df1['day_encoded']
```
