```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
df.head()
```

```python
df.info()

```

# Data cleaning

```python
df[df.isnull().any(axis=1)]
```

```python
df.loc[:122,"Region"]=0
df.loc[122:,"Region"]=1
```

```python
df.head()
```

```python
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
df.isnull().sum()
```

```python
#remove the nul values
df=df.dropna().reset_index(drop=True)
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
#Fix spaces in column names
df_columns=df.columns.str.strip()
df.columns=df_columns
df.columns
```

```python
# Change the required columns as integer data type
df[['month','day','year','Temperature','RH','Ws']] = df[['month','day','year','Temperature','RH','Ws']].astype(int) 
```

```python
df.info()
```

```python
# Changing other columns to float data type
objects=[features for features in df.columns if df[features].dtype == 'O']
for i in objects:
    if i != 'Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
# lets save the cleaned dataset
df.to_csv('Algerian_forest_fires_dataset_cleaned.csv',index=False)
```

# EDA

```python
df_copy=df.drop(['day','month','year'],axis=1)
df_copy.head()
```

```python
#Encoding of cateegorical data
df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.head()
```

```python
df_copy['Classes'].value_counts()
```

```python
#plot density plot for all features
sns.set()
df_copy.hist(bins=20,figsize=(20,15),color='blue')
plt.show()
```

```python
#percentage for pie chart
percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
#plot pie chart
classlabels=["Fire","Not Fire"]
plt.figure(figsize=(8,8))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%',startangle=90)
plt.title("Pie chart for fire and not fire classes")
plt.show()
```
