```python

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n ",series)
print(type(series))


```

```python
#Creating series from dictionary elemnts
data={"a":1,"b":2,"c":3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
#DataFrame -two dimensional
#Create dataframe form a dictionary of list
data={
    "Name":["Joy","Mercy","Neno"],
    "Age":[25,30,23],
    "City":["Nairobi","New York","Mombasa"]
}
df=pd.DataFrame(data)
print(df)
```

```python
import numpy as np
np.array(df)
```

```python
#Create dtframe form a list of dictionaries
data=[
    {"Name":"Faith","Age":20,"City":"Nairobi"},
    {"Name":"Joy","Age":22,"City":"Newyork"},
    {"Name":"Fien","Age":24,"City":"Momabsa"},
    {"Name":"Liv","Age":23,"City":"Kisumu"}
]
df=pd.DataFrame(data)
print(df)
```

```python
#Merging and joining DataFrames
import pandas as pd
df1=pd.DataFrame({'key':['A','B','C'], 'value1':[1,2,3]})
df2=pd.DataFrame({'key':['A','B','D'], 'value2':[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
#Merge the dataframes on the key columns
pd.merge(df1,df2,on='key',how='inner')
```

```python
pd.merge(df1,df2,on='key',how='outer')
```

```python
pd.merge(df1,df2,on='key',how='left')
```

```python
pd.merge(df1,df2,on='key',how='right')
```
