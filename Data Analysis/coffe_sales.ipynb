```python
import pandas as pd
df=pd.read_csv("coffe_sales.csv")
df.head()
```

```python
df.describe()
```

```python
df.dtypes
```

```python
#Handling missing values
df.isnull().sum()
```

```python
#Filling msiising values with the mean of the column
df['hour_of_day_fillNA']=df['hour_of_day'].fillna(df['hour_of_day'].mean())
df
```

```python
#Change the datatype
#convert categorical data into numerical data
df.dtypes
```

```python
#Renaming columns
df=df.rename(columns={'hour_of_day':'hour'})
df.head()
```

```python
#Change the datatype
df['money_new']=df['money'].astype(int)
df
```

```python
df['new_money_value']=df['money'].apply(lambda x:x*2)
df.head()
```

```python
#Data agrregating and grouping
grouped_mean=df.groupby('coffee_name')['hour_of_day_fillNA'].median()
print(grouped_mean)
```

```python
df.groupby(['coffee_name','Time_of_Day'])['hour_of_day_fillNA'].median()
```

```python
#Aggregate multiple functions
df.groupby(['coffee_name','Time_of_Day'])['hour_of_day_fillNA'].agg(['mean','count','median'])
```
