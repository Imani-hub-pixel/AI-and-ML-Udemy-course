```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

#create a line plot
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title(" Basic Line plot")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

#customize line plot
plt.plot(x,y,color="red",linestyle="-",marker="*",linewidth=2,markersize=9)
plt.grid(2)
```

```python
#Creating multiple plots
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

plt.figure(figsize=(9,5))
plt.subplot(1,2,1)
plt.plot(x,y1,color="green")
plt.title("plot1")


plt.subplot(1,2,2)
plt.plot(x,y2,color="green")
plt.title("plot2")


plt.subplot(2,2,3)
plt.plot(x,y2,color="red")
plt.title("plot3")

```

```python
#BAR plot

categories=['A','B','C','D','E']
values=[5,7,3,8,6]

plt.bar(categories,values,color="pink")
plt.xlabel("Categories")
plt.ylabel("values")
```

```python
#Histograms
#They are used to represnt distribution of data.They divide the data into bins and count the number of data points in each bin

data=[1,2,3,4,5,4,5,3,2,4,5,5,5,2,3,5,3,3,3]

#Create  a histogram

plt.hist(data,bins=5,color="orange",edgecolor='black')
```

```python
#Create a scatter plot
x=[1,2,3,4,5]
y=[2,3,4,5,6]

plt.scatter(x,y,color='black',marker='x')
```

```python
#Creat a pie chart-

labels=['A','B','C','D']
sizes=[20,30,40,10]
colors=('gold','yellowgreen','pink','blue')
explode=(0.2,0,0,0)#-move out the first slice

plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct='%1.1f%%',shadow=True)
```

```python
#Sales data visualization
import pandas as pd
sales_data=pd.read_csv("amazon.csv")
sales_data.head()
```

```python
sales_data.info()
```

```python
#Group based on product category
sales_data['category'] = sales_data['category'].str[:10]
sales_data['rating']=pd.to_numeric(sales_data['rating'],errors='coerce')
sales_data_by_product=sales_data.groupby('category')['rating'].sum()
print(sales_data_by_product)


```

```python

sales_data_by_product.plot(kind='bar',color='purple')
```
